### 1: A aplicação e a demonstração visual 

Rotação Horária em 90° e Cisalhamento

Utilizando dois vetores unitários, da base canônica em ℝ², sendo eles:

\begin{align*}
\vec{v_1} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} \quad
\vec{v_2} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}
\end{align*}

Juntando ambos chegasse a Matriz de vetores unitários $\vec{M_u}$:

$$
M_u = \begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
$$

E dada as Matrizes de Transformação obitidas anteriormente na Atividade 0:

\begin{align*}
R_θ &= \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix} &
C_k &= \begin{bmatrix} 1 & 3 \\ 0 & 1 \end{bmatrix}
\end{align*}

Com $R_θ$ sendo a matriz de Rotação Horária em 90° e $C_k$ sendo a matriz de Cisalhamento Horizontal

### 2: Mostrar visualmente e explicar o valor do determinante depois das transformações

##### 2.1 Dada a primeira transformação linear, que foi a de Rotação Horária, se obtém os seguintes valores em referência a área e ao valor do determinante:

```python
plot_determinante(vetores_u, matriz_ur, 'V1 e V2 após a Rotação')
```

Se pode perceber que mesmo após a aplicação da Transformação Linear de Rotação Horária o valor do Determinante e Área não foram alterados. Isso ocorre pelo fato de que a Transformação de Rotação não altera a área no plano. Além do mais, o Determinante é uma medida que serve para indicar o quanto uma transformação altera uma determinada área, nesse caso como $det = 1$ então a área não foi alterada. 

##### 2.2 A segunda transformação linear, que foi a de Rotação Cisalhamento Horizontal, originou os seguintes valores em referência a área e ao valor do determinante:

```python
plot_determinante(matriz_ur, matriz_urc, title= "V1 e V2 após Rotação e Cisalhamento")
```

Novamente, é notável que mesmo após o cisalhamento e a clara distorção que houve com o vetor $v_1$ em relação ao vetor $v_2$, a area e o Determinante permaneceram iguais aos valores da transformação anterior. Por conseguinte, da mesma forma que a Transformação de Rotação não altera a área, o mesmo ocorre com a Transformação de Cisalhamento, uma vez que ela também é uma transformação que não altera a área do plano. Por causa disso mantém-se $det = 1$ e $área = 1$ visto que ambos não se alteraram.  

### 3: Procedimento na transformação de Rotação seguida de Cisalhamento nos vetores unitários

Dada a matriz de vetores unitários $M_u$, para o procedimento ela será dividida em $v_1$ e $v_2$:

$$
M_u = \begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
$$

\begin{align*}
\vec{v_1} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} \quad
\vec{v_2} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}
\end{align*}

Elas serão utilizadas para multiplicar as matrizes de Rotação e Cisalhamento. 

##### 3.1: Vetores Unitários x Matriz de Rotação

Dada a Matriz de Rotação: 

\begin{align*}
R_θ &= \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix}
\end{align*}

A multiplicação ocorre no seguinte formato:

$$
\begin{bmatrix}
0 & 1 \\
-1 & 0
\end{bmatrix}
\cdot
\begin{pmatrix}
x \\
y
\end{pmatrix}
=
x \cdot
\begin{pmatrix}
0 \\
-1
\end{pmatrix}
+
y \cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}

=
\begin{pmatrix}
? & ? \\
? & ?
\end{pmatrix}

$$

Para o vetor $v_1$:

$$
\begin{bmatrix}
0 & 1 \\
-1 & 0
\end{bmatrix}
\cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}
=
1 \cdot
\begin{pmatrix}
0 \\
-1
\end{pmatrix}
+
0 \cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}
=
\begin{pmatrix}
0 \\
-1
\end{pmatrix}
$$



Para o vetor $v_2$:

$$
\begin{bmatrix}
0 & 1 \\
-1 & 0
\end{bmatrix}
\cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}
=
0 \cdot
\begin{pmatrix}
0 \\
-1
\end{pmatrix}
+
1 \cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}
=
\begin{pmatrix}
1 \\
0
\end{pmatrix}
$$

Juntando respectivamente os vetores e transformando em uma matriz, se tem uma Matriz Rotacionada:
$$
M_r = \begin{bmatrix}
0 & 1 \\
-1 & 0
\end{bmatrix}
$$


```python
print("Matriz Unitária Rotacionada:\n", matriz_ur)
#As colunas são exibidas como colunas deitadas, respectivamente V1 e V2
```

##### 3.2: Matriz Rotacionada x Matriz de Cisalhamento

O processo para o Cisalhamento é diferente, dada a Matriz de Cisalhamento: 
\begin{align*}
C_k &= \begin{bmatrix} 1 & 3 \\ 0 & 1 \end{bmatrix}
\end{align*}

Obtida a Matriz Rotacionada reparte-se a matriz para originar os vetores novamente $v_1$ e $v_2$:

\begin{align*}
\vec{v_1} = \begin{pmatrix} 0 \\ -1 \end{pmatrix} \quad
\vec{v_2} = \begin{pmatrix} 1 \\ 0 \end{pmatrix}
\end{align*}

A multplicação ocorre da seguinte forma: 
$$
\begin{bmatrix}
1 & 3 \\
0 & 1
\end{bmatrix}
\cdot
\begin{pmatrix}
x \\
y
\end{pmatrix}
=
x \cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}
+
y \cdot
\begin{pmatrix}
3 \\
1
\end{pmatrix}

=
\begin{pmatrix}
? & ? \\
? & ?
\end{pmatrix}

$$

Para o vetor $v_1$: 
$$
\begin{bmatrix}
1 & 3 \\
0 & 1
\end{bmatrix}
\cdot
\begin{pmatrix}
0 \\
-1
\end{pmatrix}
=
0 \cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}
+
(-1 \cdot
\begin{pmatrix}
3 \\
1
\end{pmatrix})

=
\begin{pmatrix}
-3\\
-1
\end{pmatrix}

$$


Para o vetor $v_2$:
$$
\begin{bmatrix}
1 & 3 \\
0 & 1
\end{bmatrix}
\cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}
=
1 \cdot
\begin{pmatrix}
1 \\
0
\end{pmatrix}
+
0 \cdot
\begin{pmatrix}
3 \\
1
\end{pmatrix}

=
\begin{pmatrix}
1 \\
0
\end{pmatrix}

$$


Juntando respectivamente os vetores e transformando em uma matriz resultante, se tem uma Matriz Rotacionada e Cisalhada dos vetores unitários:

$$
M_rc = \begin{bmatrix}
-3 & 1 \\
-1 & 0
\end{bmatrix}
$$

```python
print("Matriz Rotacionada e Cisalhada:\n", matriz_urc)
#As colunas são exibidas como colunas deitadas, respectivamente V1 e V2
```
