### 1: A aplicação e a demonstração visual 

Rotação Horária em 90° e Cisalhamento

Utilizando dois vetores unitários, da base canônica em ℝ², sendo eles:

\begin{align*}
\vec{V_1} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} \quad
\vec{V_2} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}
\end{align*}

Juntando ambos chegasse a Matriz de vetores unitários $\vec{M_u}$:

$$
M_u = \begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
$$

E dada as Matrizes de Transformação obitidas anteriormente na Atividade 0:

\begin{align*}
R_θ &= \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix} &
C_k &= \begin{bmatrix} 1 & 3 \\ 0 & 1 \end{bmatrix}
\end{align*}

Com $R_θ$ sendo a matriz de Rotação Horária em 90° e $C_k$ sendo a matriz de Cisalhamento Horizontal

Se pode perceber que mesmo após a aplicação da Transformação Linear de Rotação Horária o valor do Determinante e Área não foram alterados. Além do mais, o Determinante é uma medida que serve para indicar o quanto uma transformação altera uma determinada área, nesse caso como $det = 1$ então a área não foi alterada.
