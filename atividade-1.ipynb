Rotação Horária em 90° e Cisalhamento

Utilizando dois vetores unitários, da base canônica em ℝ², sendo eles:

\begin{align*}
\vec{f} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} \quad
\vec{g} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}
\end{align*}

Juntando ambos chegasse a Matriz de vetores unitários $\vec{M_u}$:

$$
M_u = \begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
$$

E dada as Matrizes de Transformação obitidas anteriormente na Atividade 0:

\begin{align*}
R_θ &= \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix} &
C_k &= \begin{bmatrix} 1 & 3 \\ 0 & 1 \end{bmatrix}
\end{align*}

Com $R_θ$ sendo a matriz de Rotação Horária em 90° e $C_k$ sendo a matriz de Cisalhamento Horizontal

```python
import numpy as np
from matplotlib import pyplot as plt    

from vetor_plot import plot_vectors
```

```python
#gerando as matrizes para matriz Unitário, matriz de rotação e matriz de cisalhamento 
vetores_u = np.array([[1, 0], [0, 1]])
matriz_r = np.array([[0, -1], [1, 0]])
matriz_c = np.array([[1, 0], [3, 1]])

matriz_ur = np.dot(vetores_u, matriz_r)
matriz_uc = np.dot(vetores_u, matriz_c)

matriz_urc = np.dot(matriz_ur, matriz_c)
matriz_ucr = np.dot(matriz_uc, matriz_r)
```
