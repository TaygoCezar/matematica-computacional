### 1 $)$ Gradiente Descendente

#### 1.1 $)$ Explicação de Conceitos Base anteriores a Descida do Gradiente

Antes de explicar propriamente Gradiente Descendente é necessário explicar alguns termos que serão abordados enquanto usa-se Gradiente Descendente. O exemplo abordado nesta atividade faz uso das variaveis peso e altura, trata-se de uma função afim de forma genérica:

##### 1.1.1 $)$ Função Linear

$f(x) = ax + b$

Com $a$ sendo o coeficiente angular(slope), ele determina se a reta será crescente($a> 0$) ou decrescente($a< 0$).

$b$ trata-se do coeficiente linear(intercept), basicamente é onde a reta da função $f(x)$ vai cruzar no eixo das ordenadas de $Y$.

No exemplo, a função dada é da Altura Prevista(Predicted Height), que basicamente é uma função linear para prever alturas baseado em pesos(a unidade de peso não tem tanta importância).

A função é da seguinte forma:

$Predicted Height = slope \cdot height + Intercept$

##### 1.1.2) Chute Inicial(Initial Guess)

O Initial Guess dá valores iniciais para os parâmetros do Gradiente Descendente antes das iterações. No caso do exemplo, os seguintes valores: 


\begin{align*}
slope = 0.64 \quad
intercept = 0
\end{align*}

Posteriormente esses valores serão usados. Não há uma forma correta para se dar um Chute Inicial, mas há formas que geram resultados mais positivos do que outras, como escolher valores pequenos ou iguais a 0. Um Initial Guess bem feito pode reduzir o número de iterações que o Gradiente Descendente teria de fazer. 

##### 1.1.3 $)$ Função de Perda(Loss Function)

uma função de perda é usada para medir o desempenho de um modelo, calculando o desvio das previsões de um modelo em relação às previsões corretas de uma base de dados. No exemplo que será seguido a Loss Function usada será a de Soma dos Quadrados dos Resíduos(Sum of the Squared Residuals). 

A função consiste na soma dos quadrados das diferenças entre os valores reais da variável dependente e os valores previstos pelo modelo.

O resíduo ao quadrado é obtido da seguinte forma:

$Residual = (Observerd Height - Predicted Height)$ <sup>2<sup>

E por tratar-se de uma soma, ela fica num somatório:

$\sum_{i=1}^{n} Residual$
