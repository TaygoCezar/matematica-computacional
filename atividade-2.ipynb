## A $)$ Descida do Gradiente

### 1 $)$ Gradiente Descendente

#### 1.1 $)$ Explicação de Conceitos Base anteriores a Descida do Gradiente

Antes de explicar propriamente Gradiente Descendente é necessário explicar alguns termos que serão abordados enquanto usa-se Gradiente Descendente. 

##### 1.1.1 $)$ Função Linear

O exemplo abordado nesta atividade faz uso das variaveis peso e altura, trata-se de uma função afim de forma genérica:

$f(x) = ax + b$

Com $a$ sendo o coeficiente angular(slope), ele determina se a reta será crescente($a> 0$) ou decrescente($a< 0$).

$b$ trata-se do coeficiente linear(intercept), basicamente é onde a reta da função $f(x)$ vai cruzar no eixo das ordenadas de $Y$.

No exemplo, a função dada é da Altura Prevista(Predicted Height), que basicamente é uma função linear para prever alturas baseado em pesos(a unidade de peso não tem tanta importância).

A função é da seguinte forma:

$Predicted Height = slope \cdot height + Intercept$

##### 1.1.2) Chute Inicial(Initial Guess)

O Initial Guess dá valores iniciais para os parâmetros do Gradiente Descendente antes das iterações. No caso do exemplo, os seguintes valores: 


\begin{align*}
slope = 0.64 \quad
intercept = 0
\end{align*}

Posteriormente esses valores serão usados. Não há uma forma correta para se dar um Chute Inicial, mas há formas que geram resultados mais positivos do que outras, como escolher valores pequenos ou iguais a 0. Um Initial Guess bem feito pode reduzir o número de iterações que o Gradiente Descendente teria de fazer. 

##### 1.1.3 $)$ Função de Perda(Loss Function)

uma função de perda é usada para medir o desempenho de um modelo, calculando o desvio das previsões de um modelo em relação às previsões corretas de uma base de dados. No exemplo que será seguido a Loss Function usada será a de Soma dos Quadrados dos Resíduos(Sum of the Squared Residuals). Também pode ser chamado de SSR.

A função consiste na soma dos quadrados das diferenças entre os valores reais da variável dependente e os valores previstos pelo modelo. As funções de perda são específicas do aprendizado supervisionado, em que as tarefas de treinamento pressupõem a existência de uma resposta correta, uma "verdade absoluta".

O resíduo ao quadrado é obtido da seguinte forma:

$Residual = (Observerd Height - Predicted Height)$ <sup>2<sup>

E por tratar-se de uma soma, ela fica num somatório:

$\sum_{i=1}^{n} Residual$



##### 1.1.4 $)$ Regra da Cadeia(Chain Rule)

A Regra da Cadeia será aplicada na Função de Perda(Loss Function) citada anteriormente a fim de determinar o Coeficiente Angular(slop) por meio de qualquer valor do Coeficiente Linear(intercept).

A Regra da Cadeia é da seguinte forma:

$$
\frac{d}{dx} f(g(x)) = f'(g(x)) \cdot g'(x)
$$

E derivando a Função de Perda é encontra-se a inclinação da reta tangente a uma curva, para a função usada ela fica no seguinte formato:

$$
\frac{d}{dIntercept} SumSquaredResiduals = -2 \cdot ( ObservedHeitght - (Intercept + 0.64 \cdot Weitght))
$$

#### 1.1.5 $)$ Passos, Taxa de Aprendizado e Novo Coeficiente Linear(New Concept)

A Descida do Gradiente dita o tamanho dos passos dados, ocorre da seguinte forma:

$StepSize = DerivadaSSR \cdot LearningRate$

O tamanho dos passos está relacioando com o Coeficiente Angular, quanto mais perto do valor ótimo, mais perto de 0 o Slope da Curva(obtida pela derivada) chega de 0. Então quanto mais próximo de 0, o valor dos passos tem de ser menor. Quanto mais longe de 0 estiver, maiores os passos podem ser, no entanto

O $Slope$ mencionado é da Derivada do algoritmo SSR.

O Novo Coeficiente Linear é obtido subtraindo o Tamanho do Passo(Step Size) do Old Intercept(Antigo Coeficiente Linear), na seguinte forma:

$New Intercept = Old Intercept - StepSize$



Esse Novo Intercept será usado na próxima iteração do Gradiente Descendente como o intercept.

#### 1.2 $)$ Explicação de Gradiente Descendente

A descida do Gradiente tem o objetivo de obter o valor mínimo da Função de Perda, isto é, o menor valor da Soma Residual ao Quadrado. Isso significa que quanto menor for a Perda(função) mais otimizado o modelo fica, mas para encontrar esse valor seriam necessários vários testes com valores de $intercept$ diferentes, o que não é nada eficiente. 

O algoritmo de Descida de Gradiente entra em jogo para tornar esse processo mais eficiente. O algoritmo inicia-se longe do valor ideal, fazendo poucos cálculos, e quanto mais perto de um valor mínimo, mais cálculos são executados. Basicamente, passos largos e passos pequenos.

#### 1.3 $)$  O código e a Animação

Para o primeiro valor de Learning Rate:

```python
#Código Refatorado

#Valores base retirados do slide
h_observada = [1.4, 1.9, 3.2]
peso = [0.5, 2.3, 2.9]

#Valores base para a Descida de Gradiente
lr = 0.1
max_steps = 100
m_step_size = 0.01

#funções

def predicted_H(slope, intercept, height):
    return slope * height + intercept

def ssr(h_observada, slope, intercept, peso):
    return sum([(h_observada[i] - predicted_H(slope, intercept, peso[i])) ** 2 for i in range(len(peso))])

def derivada_ssr(h_observada, peso, slope, intercept):
    return sum([-2 * (h_observada[i] - predicted_H(slope, intercept, peso[i])) for i in range(len(peso))])

def Gradient_Descent(slope, intercept, learn_r, max_step, minimum_step):
    slope = 0.64
    intercept = 0
    intercept_history = []

    for i in range(max_step):
        new_slope = derivada_ssr(h_observada, peso, slope, intercept)
        step_size = new_slope * learn_r
        new_intercept = intercept - step_size

        intercept = new_intercept
        intercept_history.append(intercept)

        print(f"Passo {i+1}: coeficiente angular = {new_slope:.2}, novo intercepto = {intercept:.2}, tamanho do passo = {step_size:.2}")

        if abs(step_size) < minimum_step:
            print(f"o coeficiente linear ideal foi intercept = [{intercept:.3}]")
            break
    return intercept, intercept_history

Gradient_Descent(0.64, 0, lr, max_steps, m_step_size)
```

## B $)$ Descida de Gradiente com atualização do Intercept e do Slope 

### 2 $)$ Gradiente Descendente Atualizando as duas variáveis

Diferentemente do processo anterior, neste processo ambas as váriaveis irão alterar-se conforme o algoritmo de Descida de Gradiente faz seus procedimentos internos.

No caso o Slope(Coeficiente Angular) terá o valor $slope = 1$ e o Intercept(Coeficiente Linear) terá valor 0, inicialmente, visto que ambos vão ser atualizados.

#### 2.1 $)$ Regra da Cadeia em relação ao Slope(Coeficiente Angular)

Para atualizar o slope é necessário aplicar a regra da cadeia na Soma do Quadrado dos Resíduos em relação ao  slope.

Que ficou da seguinte forma:



$$
\frac{d}{dSlope} SumSquaredResiduals = -2 \cdot Weight \cdot ( ObservedHeitght - (Intercept + Slope \cdot Weitght))
$$

#### 2.2 $)$ Step Size Slope

É criado o Step size para o slope também na seguinte forma:

$StepSizeSlope = DerivadaSSRslope \cdot LearningRate$



#### 2.3 $)$ O novo Slope

O novo Slope(Coeficiente Angular) será o seu valor Antigo menos o valor do Step Size referente ao Slope, na seguinte forma:

$New Slope = Old Slope - StepSizeSlope$

#### 2.4 $)$ O código

```python
#Código Refatorado para atualizar as duas variaveis(slope e intercept)

#Valores base retirados do slide
h_observada = [1.4, 1.9, 3.2]
peso = [0.5, 2.3, 2.9]

#Valores base para a Descida de Gradiente
lr = 0.01
max_steps = 1000
m_step_size = 0.0001

#funções

def predicted_H(slope, intercept, height):
    return slope * height + intercept

def ssr(h_observada, slope, intercept, peso):
    return sum([(h_observada[i] - predicted_H(slope, intercept, peso[i])) ** 2 for i in range(len(peso))])

def derivada_ssr(h_observada, peso, slope, intercept):
    return sum([-2 * (h_observada[i] - predicted_H(slope, intercept, peso[i])) for i in range(len(peso))])

def slope_dssr(h_observada, weight, slope, intercept):
    return sum([-2*weight[i]*(h_observada[i] - predicted_H(slope, intercept, weight[i])) for i in range(len(weight))])


def Gradient_Descent_2(slope, intercept, learn_r, max_step, minimum_step):
    slope = 1
    intercept = 0
    intercept_history = []
    slope_history = []

    for i in range(max_step):
        intercept_slope = derivada_ssr(h_observada, peso, slope, intercept)
        
        d_ssr_slope = slope_dssr(h_observada, peso, slope, intercept)
        
        step_size_intercept = intercept_slope * learn_r
        new_intercept = intercept - step_size_intercept

        step_size_slope = d_ssr_slope * learn_r
        new_slope = slope - step_size_slope

        slope = new_slope
        slope_history.append(slope)

        intercept = new_intercept
        intercept_history.append(intercept)

        if abs(step_size_intercept) < minimum_step:
            print(f"o menor coeficiente angular já foi obtido: {intercept_slope}")
            break
    return intercept, slope, intercept_history, slope_history

Gradient_Descent_2(1, 0, lr, max_steps, m_step_size)
```

### 3 $)$ Gradiente Descendente Estocástico

O Gradiente Descendente Estocástico simples, que usa apenas apenas uma amostra por passo:

```python
h_observada = [1.4, 1.9, 3.2]
peso = [0.5, 2.3, 2.9]

lr = 0.01
max_steps = 1000
m_step_size = 0.0001

def predicted_H(slop, intercepto, weight):
    return slop * weight + intercepto

def d_intercept(indices, slop, intercepto):
    return sum([-2*(h_observada[i] - predicted_H(slop, intercepto, peso[i])) for i in indices])

def d_slope(indices, slop, intercepto):
    return sum([-2*peso[i]*(h_observada[i] - predicted_H(slop, intercepto, peso[i])) for i in indices])

```

Código do Gradiente Descendente Estocástico:

```python
import random

def Gradient_Descent_Stochastic(lr, max_steps, min_step):
    slope = 1
    intercept = 0

    for step in range(max_steps):
        i = random.randint(0, 2)   # escolhe 1

        d_slop = d_slope([i], slope, intercept)
        d_intercepto = d_intercept([i], slope, intercept)

        slope = slope - lr * d_slop
        intercept = intercept - lr * d_intercepto

        if abs(lr*d_intercepto) < min_step:
            break

    return intercept, slope

intercept, slope = Gradient_Descent_Stochastic(lr, max_steps, m_step_size)

print(f" Gradiente Descendente Estocástico(1 amostra por passo):\n intercept: {intercept} \n slope: {slope}")
```

#### 3.1 $)$ Mini-Batch de 2 samples

Código do minibacht 2, no caso duas amostras por passo:

```python
def Gradient_Descent_Stochastic_MB2(lr, max_steps, min_step):
    slope = 1
    intercept = 0
    n = len(peso)

    for step in range(max_steps):
        # embaralha os índices a cada iteração
        indices = [0, 1, 2]
        random.shuffle(indices)

        # corta a lista até antes do índice 2 e pega os dois primeiros como mini-batch
        batch = indices[:2]

        d_slop = d_slope(batch, slope, intercept)
        d_intercepto = d_intercept(batch, slope, intercept)

        slope = slope - lr * d_slop
        intercept = intercept - lr * d_intercepto

        if abs(lr*d_intercepto) < min_step:
            break

    return intercept, slope

linear, angular = Gradient_Descent_Stochastic_MB2(lr, max_steps, m_step_size)

print(f" Mini-Batch 2:\n intercept: {linear} \n slope: {angular}")
```

## D $)$ Redes Neurais

### 4 $)$ Demonstração do gradiente descendente em redes neurais

#### 4.1 $)$ Bieases(Vieses)

É um valor somado após as multiplicações. Serve para deslocar a saída para cima ou para baixo, permitindo que o neurônio aprenda funções que não passam pela origem.

#### 4.2 $)$ Pesos(Weigths)
São valores que multiplicam cada entrada do neurônio. Definem a intensidade e direção da influência de cada entrada na saída.

#### 4.3 $)$ Função de Ativação(SoftPlus)
É a função aplicada à soma ponderada do neurônio $(peso × entrada + viés)$.
Ela é responsável por introduzir não linearidade no modelo, é o ponto que difere uma regressão linear de uma rede neural de verdade.

As funções de ativação são um elemento extremamente importante para Redes Neurais. Elas vão basicamente decidir se um neurônio deve ser ativado ou não. Ou seja, se a informação que o neurônio está recebendo é relevante para a informação fornecida ou deve ser ignorada.

A Softplus tem um comportamento próximo da função ReLu(função de ativação), mas com a diferença de mantém todos os neurônios ativos. Para entradas positivas grandes, o Softplus retorna valores próximos à própria entrada. Para entradas negativas, ele retorna pequenos valores positivos em vez de zero.

$$
\text{softplus}(x) = \ln(1 + e^{x})
$$


#### 4.4) A derivada de b3(viés)

Nesse primeiro caso ajusta-se o erro atualizando apenas o $b3$(viés), isso ocorre por meio da derivada da Soma ao Quadrado dos Resíduos em relação ao víes b3, que ocorre da seguinte forma:  

$$
\frac{d\,SSR}{d\,b_3}
= \frac{d\,SSR}{d\,Predicted} \times \frac{d\,Predicted}{d\,b_3}
= \sum_{i=1}^{n} -2 \, (Observed_i - Predicted_i)
$$

#### 4.5 $)$ Step Size(Passo) e o novo valor para o b3

Dado o valor do somatório da derivada em relação ao viés, se pode obter o Step Size e um novo valor para $b3$:

$StepSize = DerivadaSSR(b3) \cdot LearningRate$

E o novo valor para $b3$ é dado pela seguinte forma: 

$new(b3) = Old(b3) - StepSize$

#### 4.5 $)$ Condição de Parada

Optei por usar a condição de parada como um valor de Step Size $0.01$, chegando num valor abaixo desse a Rede Neural cessa.

#### 4.6 $)$ Código da Rede Neural com apenas o viés b3 atualizando-se.

```python
#Usando função de ativação SoftPlus
import math
#Valores
dosage = [0, 0.5, 1.0]
observed = [0, 1.0, 0]
learning_rate = 0.1

#Hidden Layers
#weigths w1 and w2
#biases b1 and b2
h_weights = [3.34, -3.53]
h_biases = [-1.43, 0.57]
#output Layer
#weight w3 and w4
#bias b3
out_weights = [-1.22, -2.30]
out_biases = 0 #inicializado em 0


#Funções

def softplus(x):
    return math.log(1 + math.exp(x))

def predict(dose, h_peso, h_vies, out_peso, out_bias):
    predicted_values = []
    for j in range(len(dose)):
        values = sum([softplus(dose[j] * h_peso[i] + h_vies[i]) * out_peso[i] for i in range(len(h_peso))])
        values += out_bias
        predicted_values.append(values)
    return predicted_values

def DSSR(observado, dose, out_bias):
    return sum([-2* (observado[i]- predict(dose, h_weights, h_biases, out_weights, out_bias)[i]) for i in range(len(observado))])

def Neural_Network(lr, dose, observed, out_bia):
    step_size = 1
    i = 0
    out_bia = 0
    while abs(step_size) > 0.01:
        predito = predict(dosage, h_weights, h_biases, out_weights, out_bia)
        dssr = DSSR(observed, dose, out_bia)
        step_size = dssr * lr
        new_biases = out_bia - step_size
        old_bia = out_bia
        out_bia = new_biases
        i += 1
        print(f"(Iteração: {i}) [Step Size: {step_size}] [Viés Antigo: {old_bia}] [Viés Novo: {new_biases}]\n")

Neural_Network(learning_rate, dosage, observed, out_biases)
    
```

## E $)$ Otimizando 3 Parâmetros Simultaneamente

#### 5 $)$ Rede Neural com 3 parâmetros sendo atualizados

Os parâmetros são $b3$, $w3$ e $w4$, cada um deles terá de ser derivado em relação a Soma do Quadrado dos Resíduos, a fim de atualizar seus valores.

#### 5.1 $)$ Processo para Atualizar o parâmetro w3

A derivação da Soma do Quadrado dos Resíduos em relação a $w3$:

$$
\frac{d\,SSR}{d\,w_3}
= \frac{d\,SSR}{d\,Predicted} \times \frac{d\,Predicted}{d\,w_3}
= \sum_{i=1}^{n} -2 \, (Observed_i - Predicted_i) \, y_{1,i}
$$


E o valor para o Step Size referente a $w3$ é o seguinte:

$StepSize(w3) = DerivadaSSR(w3) \cdot LearningRate$

O novo valor para $w3$ é obtido da seguinte forma: 

$new(w3) = Old(w3) - StepSize$

#### 5.2 $)$ Para W4, a atualização é da seguinte forma:

$$
\frac{d\,SSR}{d\,w_4}
= \frac{d\,SSR}{d\,Predicted} \times \frac{d\,Predicted}{d\,w_4}
= \sum_{i=1}^{n} -2 \, (Observed_i - Predicted_i) \, y_{1,n}
$$


E o valor para o Step Size referente a $w4$ é o seguinte:

$StepSize(w4) = DerivadaSSR(w4) \cdot LearningRate$

O novo valor para $w4$ é obtido da seguinte forma: 

$new(w4) = Old(w4) - StepSize$

#### 5.3 $)$ $y_{1,n}$ a Função de Ativação

A derivada da SSR tanto para $w3$ quanto para $w4$ adiciona o termo $y_{1,n}$ na equação, esse termo nada mais é do que os valores obtidos na função de ativação, no caso desta rede neural a Softplus.

#### 5.4 $)$ Código da Rede Neural com os 3 parâmetros otimizados

```python
b3 = 0

b12 = [-1.43, 0.57]
w12 = [3.34, -3.53]
w34 = [0.36, 0.63]

dosage = [0, 0.5, 1]
observed = [0, 1.0, 0]

#Funções
def predict(dose, h_peso, h_vies, out_peso, out_bias):
    predicted_values = []
    for j in range(len(dose)):
        values = sum([softplus(dose[j] * h_peso[i] + h_vies[i]) * out_peso[i] for i in range(len(h_peso))])
        values += out_bias
        predicted_values.append(values)
    return predicted_values

def dssrW3(dose, observed, h_peso, h_vies, out_peso, bie):
    soft_list = [softplus(dose[i] * h_peso[0] + h_vies[0]) for i in range(len(dose))] #tá correto
    preditado = predict(dose, h_peso, h_vies, out_peso, bie)

    return sum([-2 * (observed[i] - preditado[i]) * soft_list[i] for i in range(len(observed))])
    
def dssrW4(dose, observed, h_peso, h_vies, out_peso, bie):
    soft_list = [softplus(dose[i] * h_peso[1] + h_vies[1]) for i in range(len(dose))] #tá correto
    preditado = predict(dose, h_peso, h_vies, out_peso, bie)

    return sum([-2 * (observed[i] - preditado[i]) * soft_list[i] for i in range(len(observed))])

def Neural_Network_Optimized(lr, dose, observed, out_bia, h_peso, h_vies, out_peso):
    max = 100

    for i in range(max):
        step_size_w3 = dssrW3(dose, observed, h_peso, h_vies, out_peso, out_bia) * lr
        new_w3 = out_peso[0] - step_size_w3
        old_w3 = out_peso[0]
        out_peso[0] = new_w3    


        step_size_w4 = dssrW4(dose, observed, h_peso, h_vies, out_peso, out_bia) * lr
        new_w4 = out_peso[1] - step_size_w4
        old_w4 = out_peso[1]
        out_peso[1] = new_w4

        step_size_b3 = DSSR(observed, dose, out_bia) * lr
        new_b3 = out_bia - step_size_b3
        old_b3 = out_bia
        out_bia = new_b3

        print(f"[W3 Old:{old_w3}] [W3 New:{new_w3}] [Step Size W3: {step_size_w3}] [W4 Old:{old_w4:}] [W4 New:{new_w4}] [Step Size W4: {step_size_w4}] [b3 Old:{old_b3}] [b3 New:{new_b3}] [Step Size b3: {step_size_b3}] \n")


Neural_Network_Optimized(0.1, dosage, observed, b3, w12, b12, w34)
```
