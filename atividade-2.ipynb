### 1 $)$ Gradiente Descendente

#### 1.1 $)$ Explicação de Conceitos Base anteriores a Descida do Gradiente

Antes de explicar propriamente Gradiente Descendente é necessário explicar alguns termos que serão abordados enquanto usa-se Gradiente Descendente. 

##### 1.1.1 $)$ Função Linear

O exemplo abordado nesta atividade faz uso das variaveis peso e altura, trata-se de uma função afim de forma genérica:

$f(x) = ax + b$

Com $a$ sendo o coeficiente angular(slope), ele determina se a reta será crescente($a> 0$) ou decrescente($a< 0$).

$b$ trata-se do coeficiente linear(intercept), basicamente é onde a reta da função $f(x)$ vai cruzar no eixo das ordenadas de $Y$.

No exemplo, a função dada é da Altura Prevista(Predicted Height), que basicamente é uma função linear para prever alturas baseado em pesos(a unidade de peso não tem tanta importância).

A função é da seguinte forma:

$Predicted Height = slope \cdot height + Intercept$

##### 1.1.2) Chute Inicial(Initial Guess)

O Initial Guess dá valores iniciais para os parâmetros do Gradiente Descendente antes das iterações. No caso do exemplo, os seguintes valores: 


\begin{align*}
slope = 0.64 \quad
intercept = 0
\end{align*}

Posteriormente esses valores serão usados. Não há uma forma correta para se dar um Chute Inicial, mas há formas que geram resultados mais positivos do que outras, como escolher valores pequenos ou iguais a 0. Um Initial Guess bem feito pode reduzir o número de iterações que o Gradiente Descendente teria de fazer. 

##### 1.1.3 $)$ Função de Perda(Loss Function)

uma função de perda é usada para medir o desempenho de um modelo, calculando o desvio das previsões de um modelo em relação às previsões corretas de uma base de dados. No exemplo que será seguido a Loss Function usada será a de Soma dos Quadrados dos Resíduos(Sum of the Squared Residuals). Também pode ser chamado de SSR.

A função consiste na soma dos quadrados das diferenças entre os valores reais da variável dependente e os valores previstos pelo modelo. As funções de perda são específicas do aprendizado supervisionado, em que as tarefas de treinamento pressupõem a existência de uma resposta correta, uma "verdade absoluta".

O resíduo ao quadrado é obtido da seguinte forma:

$Residual = (Observerd Height - Predicted Height)$ <sup>2<sup>

E por tratar-se de uma soma, ela fica num somatório:

$\sum_{i=1}^{n} Residual$



##### 1.1.4 $)$ Regra da Cadeia(Chain Rule)

A Regra da Cadeia será aplicada na Função de Perda(Loss Function) citada anteriormente a fim de determinar o Coeficiente Angular(slop) por meio de qualquer valor do Coeficiente Linear(intercept).

A Regra da Cadeia é da seguinte forma:

$$
\frac{d}{dx} f(g(x)) = f'(g(x)) \cdot g'(x)
$$

E derivando a Função de Perda é encontra-se a inclinação da reta tangente a uma curva, para a função usada ela fica no seguinte formato:



$$
\frac{d}{dx} SumSquaredResiduals = -2 \cdot ( ObservedHeitght - (Intercept + 0.64 \cdot Weitght))
$$

#### 1.1.5 $)$ Passos, Taxa de Aprendizado e Novo Coeficiente Linear(New Concept)

A Descida do Gradiente dita o tamanho dos passos dados, ocorre da seguinte forma:

$StepSize = Slope \cdot LearningRate$

O tamanho dos passos está relacioando com o Coeficiente Angular, quanto mais perto do valor ótimo, mais perto de 0 o Slope da Curva(obtida pela derivada) chega de 0. Então quanto mais próximo de 0, o valor dos passos tem de ser menor. Quanto mais longe de 0 estiver, maiores os passos podem ser, no entanto

O $Slope$ mencionado é da Derivada do algoritmo SSR.

O Novo Coeficiente Linear é obtido subtraindo o Tamanho do Passo(Step Size) do Old Intercept(Antigo Coeficiente Linear), na seguinte forma:

$New Intercept = Old Intercept - StepSize$



Esse Novo Intercept será usado na próxima iteração do Gradiente Descendente como o intercept.

#### 1.2 $)$ Explicação de Gradiente Descendente

A descida do Gradiente tem o objetivo de obter o valor mínimo da Função de Perda, isto é, o menor valor da Soma Residual ao Quadrado. Isso significa que quanto menor for a Perda(função) mais otimizado o modelo fica, mas para encontrar esse valor seriam necessários vários testes com valores de $intercept$ diferentes, o que não é nada eficiente. 

O algoritmo de Descida de Gradiente entra em jogo para tornar esse processo mais eficiente. O algoritmo inicia-se longe do valor ideal, fazendo poucos cálculos, e quanto mais perto de um valor mínimo, mais cálculos são executados. Basicamente, passos largos e passos pequenos.

```python
#Código Refatorado

#Valores base retirados do slide
h_observada = [1.4, 1.9, 3.2]
peso = [0.5, 2.3, 2.9]

#Valores base para a Descida de Gradiente
lr = 0.1
max_steps = 100
m_step_size = 0.01

#funções

def predicted_H(slope, intercept, height):
    return slope * height + intercept

def ssr(h_observada, slope, intercept, peso):
    return sum([(h_observada[i] - predicted_H(slope, intercept, peso[i])) ** 2 for i in range(len(peso))])

def derivada_ssr(h_observada, peso, slope, intercept):
    return sum([-2 * (h_observada[i] - predicted_H(slope, intercept, peso[i])) for i in range(len(peso))])

def Gradient_Descent(slope, intercept, learn_r, max_step, minimum_step):
    slope = 0.64
    intercept = 0
    intercept_history = []

    for i in range(max_step):
        new_slope = derivada_ssr(h_observada, peso, slope, intercept)
        step_size = new_slope * learn_r
        new_intercept = intercept - step_size

        intercept = new_intercept
        intercept_history.append(intercept)

        print(f"Passo {i+1}: coeficiente angular = {new_slope:.2}, novo intercepto = {intercept:.2}, tamanho do passo = {step_size:.2}")

        if abs(step_size) < minimum_step:
            print(f"o menor coeficiente angular já foi obtido: {new_slope}")
            break
    return intercept, intercept_history

Gradient_Descent(0.64, 0, lr, max_steps, m_step_size)
```
