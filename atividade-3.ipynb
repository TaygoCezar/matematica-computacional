### Método da Bissecção

#### Explicação do Método da Bissecção:

Dada uma função $f(x)$ ela tem de ser contínua num intervalo $[a,b]$, de forma que $f(a)\cdot f(b) < 0$.

O objetivo do método da Bissecção é diminuir a abrangência do dado intervalo que contém uma raíz até alcançar uma precisão determinada.   
Se assume que o intervalo dado tenha apenas uma raíz para a dada função $f(x) = 0$.  

Adiante, isso ocorre da seguinte forma: $ (b-a) < \epsilon$, essa condição é testada múltiplas vezes por meio da divisão iterativa do intervalo ao meio. 

O valor de a pode aumentar e o de b pode diminuir, a fim de diminuir o intervalo, logicamente restringindo o intervalo de valores onde a raíz pode estar. O método não é exato, em razão disso existe o $\epsilon$ que determina uma precisão, que seria um valor de intervalo aceitável onde a raíz pode estar situada.  
Esse método tem a desvantagem de ter uma convergência lenta para problemas mais complexos, que exigem um alto número de iterações para chegar em um resultado aceitável.

Complexidade do Método da Bissecção:  
A estimativa do número de iterações, partindo do presuposto que $k$ satisfaz a seguinte relação:

$$ k > \frac{log(b_0 - a_0) - log(\epsilon)}{log(2)} $$

Os logaritmos usam base 10, então $log(2) = 0.3010$.  
Essa estimativa serve para calcular o mínimo de iterações necessárias para encontrar a raíz da função, antes mesmo de efetuar o cálculo do problema.
  

```python
# Dado um intervalo inicial [1,2] 
def Estimativa_Bisseccao(intervalo, precisao):
    a = intervalo[0]
    b = intervalo[1]

    estimativa = (math.log10(b-a) - math.log10(precisao)/ math.log10(2))
    print(f"O número provável de iterações é: ({round(estimativa)})") 

Estimativa_Bisseccao([1,2], 10**(-6))
```

Código para o Método da Bissecção

```python
# Algoritmo 1
# Código do Método da Bisseção

# função para teste
def funcao(x):
    return x**3 - x - 1                                               
                                               # Linha 1: Os dados iniciais                                              
interval = [1.0, 2.0]                          # O intervalo Inicial [a,b]
e = 10**(-6)                                   # precisão ε

# função propriamente

def bisseccao(func, interval, e, p):
    k = 0
    a = interval[0]
    b = interval[1]
    total_i = []

    if abs(b-a) < e:                           # Linha 2: Condição para parada antes de iniciar a função, optei por usar o valor central entre o intervalo
        xi = (a+b)/2
        return xi

    k = 1                                      # Linha 3: Introdução de um iterador na função

    while abs(b-a) > e:
        i = 0
        i += 3

        M = func(a)
        i += 2                                 # Linha 4: variável para armazenar o valor de f(a) 
        x = (a + b)/2                          # Linha 5: Média da distância entre os intervalos
        i += 3

        if (func(x)*M) > 0:                    # Linha 6: Condição para que a = x
            i += 3
            a = x  
            i += 1
            if (b - a) < e:                    # Linha 8: Condição de parada
                i += 2
                xi = (a+b)/2
                i += 3
                total_i.append(i)              # Adiciona todos as operações a uma lista e tira a média do total de operações por iteração
                media_i = round((sum(total_i))/k)    
                                               
                if p == 1:                     # Print com Tabulate
                    resultados = [ ["Método", "Bissecção"], ["Intervalo inicial", f"[{interval[0]}, {interval[1]}]"], ["x̅ (raiz aproximada)", f"{xi:.6f}"], ["f(x̅)", f"{func(x):.13f}"], ["Erro x̅", f"{(b-a):.13f}"], ["Iterações", k], ["Operações(M)", media_i], ["Operação Total", sum(total_i)]]
                    print("\n" + "="*42+"\n            MÉTODO DA BISSECÇÃO\n"+"="*42)
                    print(tabulate(resultados, tablefmt="fancy_grid")) 
                    print("="*42 + "\n")       # O retorno é o x̅, o número de iterações, o valor da função em x e o erro em x(b-a)   
                    break
                else:
                    return x, k, func(x), (b-a)
                
            else:
                k += 1                         # Linha 9: Aumento da contagem do Iterador
                total_i.append(i)
        else: 
            k += 1                             # Linha 9: Aumento da contagem do Iterador
            b = x                              # Linha 7: b = x
            i += 1
            total_i.append(i)
                                               # Adição de contadores para o número de 
```

### Método 2: Método da Posição Falsa

#### Explicação do Método da Posição Falsa

Dada uma função $f(x)$ e um intervalo $[a,b]$, de forma que $f(a)\cdot f(b) < 0$.

Se pressupõe que existe apenas uma raíz no intervalo dado, então $f(x)=0$.  
O método da posição falsa não usa simplesmente uma média aritmética $\frac{a+b}{2}$, ele executa uma média aritmética ponderada entre $a$ e $b$ com os pesos $|f(a)|$ e $|f(b)|$, na seguinte forma:

$$ x = \frac{a \cdot |f(b)| + b \cdot |f(a)|}{|f(b)| + |f(a)|} = \frac {a \cdot f(b) - b \cdot f(a)}{f(b) - f(a)}$$

Complexidade do Métoda da Posição Falsa:  
Diferente da Bissecção, a Posição Falsa não possuí uma fórmula para estimar seu número de iterações, visto que a redução de seu intervalo não ocorre de maneira uniforme e depende da função que está sendo usada. Além do mais, se for exigido que dois critérios de parada sejam atendidos simultaneamente, o método pode exceder um número máximo de iterações.  
Então sua complexidade pode ser medida como o número de iterações obtidas ao fim da execução do método, além do fato de que o número de iterações vai depender da função a qual está sendo usada nele.

### Método 3: Método do Ponto Fixo(MPF)

#### Explicação do Método do Ponto Fixo(MPF)

Dada uma função contínua $f(x)$ num intervalo $[a,b]$, esse intervalo detém uma raíz da equação $f(x) = 0$.  
O método baseia-se em converter essa equação numa função equivalente $x = $ φ $(x)$ e por meio de um $x_0$ inicial gerar uma sequência 
$ \{x_k \}$ para $\xi$ usando a relação $x_{k+1} = $ φ$(x_{k})$, em razão da função φ$(x)$ ser tal que $f(\xi) = 0$ para que φ $(xi) = \xi $. Dessa forma o problema torna-se encontrar um ponto fixo em φ$(x)$.

  <img src="assets/atv_3/MPF_a.png" width="500">
</div>

```python
# Algoritmo 3
def funcX(x):
    return x**3 - x - 1

def funcG(x):
    return (x+1)**(1/3)
                                                            # Linha 1: Os dados iniciais                                                    
x0 = 1                                                      # (a): Valor Inicial de x0
e1 = 10 **(-6)                                              # (b): Dados de Precisão


# Função Propriamente
def MPF(x_0, e1, e2, fx, fg, p):
    total_i = []
    dado_x = x_0

    if abs(fx(x_0)) < e1:                                   # Linha 2: Condição de parada inicial do MPF
        print(x_0)
        xi = x_0
        return xi, 0, fx(x_0), 0
    
    k = 1                                                   # Linha 3: Definição da variável de iteração 
    while abs(fx(x_0)) > e1:    
        i = 0
        i += 2
        x_1 = fg(x_0)                                       # Linha 4: Alteração do valor de x_1 baseado no valor de gx(x_0) 
        i += 2

        if abs(fx(x_1)) < e1 or abs(x_1 - x_0) < e2:        # Linha 5: Dada a condição x̅ = x_1
            i += 2
            xi = x_1
            i += 1                
            total_i.append(i)   

            if p == 1:
                resultados = [ ["Método", "MPF"], ["Dados Iniciais", f"X₀ = {dado_x}"], ["x̅ (raiz aproximada)", f"{xi:.6f}"], ["f(x̅)", f"{fx(xi):.13f}"], ["Erro x̅", f"{(abs(x_1 - x_0)) / abs(x_1):.13f}"], ["Iterações", k], ["Operações(M)",  round((sum(total_i)/k))], ["Operação Total", sum(total_i)]]
                print("\n" + "="*42+"\n               MÉTODO DA MPF\n"+"="*42)
                print(tabulate(resultados, tablefmt="fancy_grid")) 
                print("="*42 + "\n")        
                break
            else:  
                return xi, k, fx(xi), (abs(x_1 - x_0)) / abs(x_1), round((sum(total_i)/k)), sum(total_i) 
        
        else:
            k += 1                                          # Linha 6: Adição no iterador
            x_0 = x_1                                       # Linha 7: Atualização do valor de x_0
            i += 1
            total_i.append(i)
```

```python
# teste
MPF(x0, e1, e1, funcX, funcG, 1)
```
