### Método da Bissecção

#### Explicação do Método da Bissecção:

Dada uma função $f(x)$ ela tem de ser contínua num intervalo $[a,b]$, de forma que $f(a)\cdot f(b) < 0$.

O objetivo do método da Bissecção é diminuir a abrangência do dado intervalo que contém uma raíz até alcançar uma precisão determinada.   
Se assume que o intervalo dado tenha apenas uma raíz para a dada função $f(x) = 0$.  

Adiante, isso ocorre da seguinte forma: $ (b-a) < \epsilon$, essa condição é testada múltiplas vezes por meio da divisão iterativa do intervalo ao meio. 

O valor de a pode aumentar e o de b pode diminuir, a fim de diminuir o intervalo, logicamente restringindo o intervalo de valores onde a raíz pode estar. O método não é exato, em razão disso existe o $\epsilon$ que determina uma precisão, que seria um valor de intervalo aceitável onde a raíz pode estar situada.  
Esse método tem a desvantagem de ter uma convergência lenta para problemas mais complexos, que exigem um alto número de iterações para chegar em um resultado aceitável.

Complexidade do Método da Bissecção:  
A estimativa do número de iterações, partindo do presuposto que $k$ satisfaz a seguinte relação:

$$ k > \frac{log(b_0 - a_0) - log(\epsilon)}{log(2)} $$

Os logaritmos usam base 10, então $log(2) = 0.3010$.  
Essa estimativa serve para calcular o mínimo de iterações necessárias para encontrar a raíz da função, antes mesmo de efetuar o cálculo do problema.
  

```python
# Dado um intervalo inicial [1,2] 
def Estimativa_Bisseccao(intervalo, precisao):
    a = intervalo[0]
    b = intervalo[1]

    estimativa = (math.log10(b-a) - math.log10(precisao)/ math.log10(2))
    print(f"O número provável de iterações é: ({round(estimativa)})") 

Estimativa_Bisseccao([1,2], 10**(-6))
```

### Método 2: Método da Posição Falsa

#### Explicação do Método da Posição Falsa

Dada uma função $f(x)$ e um intervalo $[a,b]$, de forma que $f(a)\cdot f(b) < 0$.

Se pressupõe que existe apenas uma raíz no intervalo dado, então $f(x)=0$.  
O método da posição falsa não usa simplesmente uma média aritmética $\frac{a+b}{2}$, ele executa uma média aritmética ponderada entre $a$ e $b$ com os pesos $|f(a)|$ e $|f(b)|$, na seguinte forma:

$$ x = \frac{a \cdot |f(b)| + b \cdot |f(a)|}{|f(b)| + |f(a)|} = \frac {a \cdot f(b) - b \cdot f(a)}{f(b) - f(a)}$$

</div>

Complexidade do Métoda da Posição Falsa:  
Diferente da Bissecção, a Posição Falsa não possuí uma fórmula para estimar seu número de iterações, visto que a redução de seu intervalo não ocorre de maneira uniforme e depende da função que está sendo usada. Além do mais, se for exigido que dois critérios de parada sejam atendidos simultaneamente, o método pode exceder um número máximo de iterações.  
Então sua complexidade pode ser medida como o número de iterações obtidas ao fim da execução do método, além do fato de que o número de iterações vai depender da função a qual está sendo usada nele. 


Código para o Método da Posição Falsa:

```python
# Algoritmo 2 
# Código do Método da Posição Falsa

# função para teste
def funcao(x):
    return x**3 - x - 1                                                                                 
                                                                                     # Linha 1: Os dados iniciais                                              
interval = [1.0, 2.0]                                                                # O intervalo Inicial [a,b]
e1 = 10**(-6)
e2 = e1                                                                              # precisão ε

# função propriamente

def PosicaoFalsa(func, interval, e1, e2, p):
    k = 0
    a = interval[0]
    b = interval[1]
    total_i = []
    tempo_t = []

    if abs(b-a) < e1:                                                                 # Linha 2: Condição para parada antes de iniciar a função, optei por usar o valor central entre o intervalo
        xi = (a+b)/2
        return {"Método": "Secante", "Dados Iniciais": f"[{interval[0]}, {interval[1]}]", "Raíz aproximada x̅": f"{xi:.6f}", "F(x̅)": f"{func(xi):.13f}", "Erro em X": 0, "N° de Iterações": 0, "Operações/Iteração": 0, "Total de Operações": 0, "Tempo Total": 0, "Tempo/Iteração": 0}
        # return xi

    k = 1                                                                            # Linha 3: Introdução de um iterador na função

    while abs(b-a) > e:
        time_inicial = time.perf_counter()
        i = 0

        M = func(a)                                                                  # Linha 4: variável para armazenar o valor de f(a) 
        x = (a*(func(b)) - b*(func(a))) / (func(b) - func(a))                        # Linha 5: Média ponderada da distância entre os intervalos 
        
        i += 11

        if abs(func(x)) < e2:
            i += 2
            total_i.append(i)
            media_i = round(sum(total_i)/k)
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
            error = abs(b-a)
            if p == 1:
                resultados = [ ["Método", "Posição Falsa"], ["Dados Iniciais", f"[{interval[0]}, {interval[1]}]"], ["x̅ (raiz aproximada)", f"{x:.6f}"], ["f(x̅)", f"{func(x):.13f}"], ["Erro x̅", f"{error:.7f}"], ["Iterações", k], ["Operação(M)", media_i], ["Operação Total", sum(total_i)], ["Tempo Total", f"{sum(tempo_t):.12f}s"], ["Tempo por Iteração", f"{(sum(tempo_t) / k):.12f}s"]]
                print("\n" + "="*42+"\n          MÉTODO DA POSIÇÃO FALSA\n"+"="*42)
                print(tabulate(resultados, tablefmt="fancy_grid")) 
                print("="*42 + "\n")                                                 # O retorno é o x̅, o número de iterações, o valor da função em x e o erro em x(b-a)   
                break      
            else:
                return {"Método": "Posição Falsa", "Dados Iniciais": f"[{interval[0]}, {interval[1]}]","Raíz aproximada x̅": f"{x:.6f}", "F(x̅)": f"{func(x):.13f}", "Erro em X": f"{(b-a):.13f}", "N° de Iterações": k, "Operações/Iteração": media_i, "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"}
                # return x, k, func(x), (b-a)       

        if (func(x)*M) > 0:                                                          # Linha 6: Condição para que a = x
            a = x 
            i += 4
            if (b - a) < e1:                                                         # Linha 8: Condição de parada
                xi = (a+b)/2
                i += 5                                                               # Print com Tabulate
                total_i.append(i)
                media_i = round(sum(total_i)/k)
                time_final = time.perf_counter()
                tempo_t.append(time_final - time_inicial)
                error = abs(b-a)
                if p == 1:
                    resultados = [ ["Método", "Posição Falsa"], ["Dados Iniciais", f"[{interval[0]}, {interval[1]}]"], ["x̅ (raiz aproximada)", f"{x:.6f}"], ["f(x̅)", f"{func(x):.13f}"], ["Erro x̅", f"{error:.7f}"], ["Iterações", k], ["Operação(M)", media_i], ["Operação Total", sum(total_i)]]
                    print("\n" + "="*42+"\n          MÉTODO DA POSIÇÃO FALSA\n"+"="*42)
                    print(tabulate(resultados, tablefmt="fancy_grid")) 
                    print("="*42 + "\n")                                             # O retorno é o x̅, o número de iterações, o valor da função em x e o erro em x(b-a)   
                    break              
                else: 
                    return {"Método": "Secante", "Dados Iniciais": f"[{interval[0]},{interval[1]}]", "Raíz aproximada x̅": f"{x:.6f}", "F(x̅)": f"{func(x):.13f}", "Erro em X": f"{(b-a):.13f}", "N° de Iterações": k, "Operações/Iteração": media_i, "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"}                      

            else:
                total_i.append(i)
                k += 1                                                               # Linha 9: Aumento da contagem do Iterador
                time_final = time.perf_counter()
                tempo_t.append(time_final - time_inicial)
        else: 
            k += 1                                                                   # Linha 9: Aumento da contagem do Iterador
            b = x 
            i += 1
            total_i.append(i)                                                                   # Linha 7: b = x
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
```

### Método 3: Método do Ponto Fixo(MPF)

#### Explicação do Método do Ponto Fixo(MPF)

Dada uma função contínua $f(x)$ num intervalo $[a,b]$, esse intervalo detém uma raíz da equação $f(x) = 0$.  
O método baseia-se em converter essa equação numa função equivalente $x = $ φ $(x)$ e por meio de um $x_0$ inicial gerar uma sequência 
$ \{x_k \}$ para $\xi$ usando a relação $x_{k+1} = $ φ$(x_{k})$, em razão da função φ$(x)$ ser tal que $f(\xi) = 0$ para que φ $(xi) = \xi $. Dessa forma o problema torna-se encontrar um ponto fixo em φ$(x)$.

</div>

Complexidade do Método do Ponto Fixo:


```python
# Algoritmo 3
def funcX(x):
    return x**3 - x - 1

def funcG(x):
    return (x+1)**(1/3)
                                                            # Linha 1: Os dados iniciais                                                    
x0 = 1                                                      # (a): Valor Inicial de x0
e1 = 10 **(-6)                                              # (b): Dados de Precisão


# Função Propriamente
def MPF(x_0, e1, e2, fx, fg, p):
    total_i = []
    dado_x = x_0
    tempo_t = []

    if abs(fx(x_0)) < e1:                                   # Linha 2: Condição de parada inicial do MPF
        print(x_0)
        xi = x_0
        return {"Método": "", "Dados Iniciais": f"X₀ = {dado_x}", "Raíz aproximada x̅": f"{xi:.6f}", "F(x̅)": f"{fx(xi):.13f}", "Erro em X": 0, "N° de Iterações": 0, "Operações/Iteração": 0, "Total de Operações": 0, "Tempo Total": 0, "Tempo/Iteração": 0}
        # return xi, 0, fx(x_0), 0
    
    k = 1                                                   # Linha 3: Definição da variável de iteração 
    while abs(fx(x_0)) > e1:    
        time_inicial = time.perf_counter()
        i = 0
        x_1 = fg(x_0)                                       # Linha 4: Alteração do valor de x_1 baseado no valor de gx(x_0) 
        i += 4

        if abs(fx(x_1)) < e1 or abs(x_1 - x_0) < e2:        # Linha 5: Dada a condição x̅ = x_1
            xi = x_1
            i += 3                
            total_i.append(i)   
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
            error = abs(x_1 - x_0)

            if p == 1:                                                                                                                                    # (abs(x_1 - x_0)) / abs(x_1)
                resultados = [ ["Método", "MPF"], ["Dados Iniciais", f"X₀ = {dado_x}"], ["x̅ (raiz aproximada)", f"{xi:.6f}"], ["f(x̅)", f"{fx(xi):.13f}"], ["Erro x̅", f"{error:.13f}"], ["Iterações", k], ["Operações(M)",  round((sum(total_i)/k))], ["Operação Total", sum(total_i)], ["Tempo Total", f"{sum(tempo_t):.12f}s"], ["Tempo por Iteração", f"{(sum(tempo_t) / k):.12f}s"]]
                print("\n" + "="*42+"\n               MÉTODO DA MPF\n"+"="*42)
                print(tabulate(resultados, tablefmt="fancy_grid")) 
                print("="*42 + "\n")        
                break
            else:  
                return {"Método": "MPF", "Dados Iniciais": f"X₀ = {dado_x}", "Raíz aproximada x̅": f"{xi:.6f}", "F(x̅)": f"{fx(xi):.13f}", "Erro em X": f"{error:.13f}", "N° de Iterações": k, "Operações/Iteração": round((sum(total_i)/k)), "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"} 
                # return xi, k, fx(xi), (abs(x_1 - x_0)) / abs(x_1), round((sum(total_i)/k)), sum(total_i)
        
        else:
            k += 1                                          # Linha 6: Adição no iterador
            x_0 = x_1                                       # Linha 7: Atualização do valor de x_0
            i += 1
            total_i.append(i)
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
```

```python
# Algoritmo 5

def f(x):
    return x**(3) - x - 1

                                                    # Linha 1: Dados iniciais

e1 = 10 ** (-6)                                     # (b): Precisões e1 e e2
e2 = e1

# Função propriamente
def Secante(x_0, x_1, e1, e2, f, p):
    total_i = []
    tempo_t = []
    dado_x0 = x_0
    dado_x1 = x_1

    if abs(f(x_0)) < e1:                            # Linha 2: Condição de parada inicial, verifica antes de entrar no loop
        xi = x_0
        return {"Método": "", "Dados Iniciais": f"X₀ = {dado_x0}, X₁ = {dado_x1}", "Raíz aproximada x̅": f"{xi:.6f}", "F(x̅)": f"{f(xi):.13f}", "Erro em X": 0, "N° de Iterações": 0, "Operações/Iteração": 0, "Total de Operações": 0, "Tempo Total": 0, "Tempo/Iteração": 0}
        # return xi
    
    if abs(f(x_1)) < e1 or abs(x_1 - x_0) < e2:     # Linha 3: Segunda condição de parada retornado x_1 dessa vez
        xi = x_1
        return {"Método": "", "Dados Iniciais": f"X₀ = {dado_x0}, X₁ = {dado_x1}", "Raíz aproximada x̅": f"{xi:.6f}", "F(x̅)": f"{f(xi):.13f}", "Erro em X": 0, "N° de Iterações": 0, "Operações/Iteração": 0, "Total de Operações": 0, "Tempo Total": 0, "Tempo/Iteração": 0}
        # return xi

    k = 1                                           # Linha 4: Inicialização do iterador
    while abs(f(x_0)) > e1:
        time_inicial = time.perf_counter()
        i = 0
        x_2 = x_1 - ( f(x_1) / (f(x_1) - f(x_0) ) ) * (x_1 - x_0) # Linha 5: x_2 recebendo seu primeiro valor baseado em x_0 e x_1
        i += 8

        if abs(f(x_2)) < e1 or abs(x_2 - x_1) < e2: # Linha 6: Condição de parada utilizando o valor de x_2
            xi = x_2
            i += 3
            total_i.append(i)
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
            error = abs(x_2 - x_1)
            if p == 1:                                                                                                                                                        #((abs(x_2 - x_1)) / abs(x_2))
                resultados = [ ["Método", "Secante"], ["Dados Iniciais", f"X₀ = {dado_x0}, X₁ = {dado_x1}"], ["x̅ (raiz aproximada)", f"{xi:.6f}"], ["f(x̅)", f"{f(xi):.15f}"], ["Erro x̅", f"{error:.13f}"], ["Iterações", k], ["Operações(M)",  round((sum(total_i)/k))], ["Operação Total", sum(total_i)], ["Tempo Total", f"{sum(tempo_t):.12f}s"], ["Tempo por Iteração", f"{(sum(tempo_t) / k):.12f}s"]]
                print("\n" + "="*42+"\n           MÉTODO DA SECANTE\n"+"="*42)
                print(tabulate(resultados, tablefmt="fancy_grid")) 
                print("="*42 + "\n")        
                break 
            else:
                return {"Método": "Newton-Haphson", "Dados Iniciais": f"X₀ = {dado_x0}, X₁ = {dado_x1}", "Raíz aproximada x̅": f"{xi:.6f}", "F(x̅)": f"{f(xi):.15f}", "Erro em X": f"{((abs(x_2 - x_1)) / abs(x_2)) :.13f}", "N° de Iterações": k, "Operações/Iteração": round((sum(total_i)/k)), "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"}
                # return xi, f(x_2), k
        else:
            x_0 = x_1                               # Linha 7: Atualização de x_0 e x_1
            x_1 = x_2   
            k += 1                                  # Linha 8: Atualização do iterador
            i += 2                            
            total_i.append(i)
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)

```
