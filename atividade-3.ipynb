### Método da Bissecção

#### Explicação do Método da Bissecção:

Dada uma função $f(x)$ ela tem de ser contínua num intervalo $[a,b]$, de forma que $f(a)\cdot f(b) < 0$.

O objetivo do método da Bissecção é diminuir a abrangência do dado intervalo que contém uma raíz até alcançar uma precisão determinada.   
Se assume que o intervalo dado tenha apenas uma raíz para a dada função $f(x) = 0$.  

Adiante, isso ocorre da seguinte forma: $ (b-a) < \epsilon$, essa condição é testada múltiplas vezes por meio da divisão iterativa do intervalo ao meio. 

O valor de a pode aumentar e o de b pode diminuir, a fim de diminuir o intervalo, logicamente restringindo o intervalo de valores onde a raíz pode estar. O método não é exato, em razão disso existe o $\epsilon$ que determina uma precisão, que seria um valor de intervalo aceitável onde a raíz pode estar situada.  
Esse método tem a desvantagem de ter uma convergência lenta para problemas mais complexos, que exigem um alto número de iterações para chegar em um resultado aceitável.

Complexidade do Método da Bissecção:  
A estimativa do número de iterações, partindo do presuposto que $k$ satisfaz a seguinte relação:

$$ k > \frac{log(b_0 - a_0) - log(\epsilon)}{log(2)} $$

Os logaritmos usam base 10, então $log(2) = 0.3010$.  
Essa estimativa serve para calcular o mínimo de iterações necessárias para encontrar a raíz da função, antes mesmo de efetuar o cálculo do problema.
  

```python
# Dado um intervalo inicial [1,2] 
def Estimativa_Bisseccao(intervalo, precisao):
    a = intervalo[0]
    b = intervalo[1]

    estimativa = (math.log10(b-a) - math.log10(precisao)/ math.log10(2))
    print(f"O número provável de iterações é: ({round(estimativa)})") 

Estimativa_Bisseccao([1,2], 10**(-6))
```

Código para o Método da Bissecção

```python
# Algoritmo 1
# Código do Método da Bisseção

# função para teste
def funcao(x):
    return x**3 - x - 1                                               
                                               # Linha 1: Os dados iniciais                                              
interval = [1.0, 2.0]                          # O intervalo Inicial [a,b]
e = 10**(-6)                                   # precisão ε

# função propriamente

def bisseccao(func, interv, e, p):
    k = 0
    a = interv[0]
    b = interv[1]
    total_i = []
    tempo_t = []
    logical_i = 0

    if abs(b-a) < e:                           # Linha 2: Condição para parada antes de iniciar a função, optei por usar o valor central entre o intervalo
        xi = (a+b)/2
        logical_i += 1
        return {"Método": "Bissecção", "Raíz x̅": f"{xi:.8f}", "F(x̅)": f"{func(xi):.13f}", "Erro em X": 0, "N° de Iterações": 0, "N° Lógico": logical_i, "Operações/Iteração": 0, "Total de Operações": 0, "Tempo Total": 0, "Tempo/Iteração": 0}
        # return xi

    k = 1                                      # Linha 3: Introdução de um iterador na função

    while abs(b-a) > e:
        time_inicial = time.perf_counter()
        i = 0

        M = func(a)                            # Linha 4: variável para armazenar o valor de f(a)                                  
        x = (a + b)/2                          # Linha 5: Média da distância entre os intervalos
        i += 8

        if (func(x)*M) > 0:                    # Linha 6: Condição para que a = x
            logical_i += 1
            a = x  
            i += 4
            if (b - a) < e:                    # Linha 8: Condição de parada
                logical_i += 1
                xi = (a+b)/2
                i += 5
                total_i.append(i)              # Adiciona todos as operações a uma lista e tira a média do total de operações por iteração
                media_i = round((sum(total_i))/k)    
                time_final = time.perf_counter()
                tempo_t.append(time_final - time_inicial)    
                error = abs(b-a)                         
                if p == 1:                     # Print com Tabulate
                    resultados = [ ["Método", "Bissecção"], ["Dados Iniciais", f"[{interv[0]}, {interv[1]}]"], ["x̅ (raiz aproximada)", f"{xi:.8f}"], ["f(x̅)", f"{func(x):.13f}"], ["Erro x̅", f"{error:.13f}"], ["Iterações", k], ["Operações Média", media_i], ["Operação Total", sum(total_i)], ["Tempo Total", f"{sum(tempo_t):.12f}s"], ["Tempo por Iteração", f"{(sum(tempo_t) / k):.12f}s"]]
                    print("\n" + "="*42+"\n            MÉTODO DA BISSECÇÃO\n"+"="*42)
                    print(tabulate(resultados, tablefmt="fancy_grid")) 
                    print("="*42 + "\n")       # O retorno é o x̅, o número de iterações, o valor da função em x e o erro em x(b-a)   
                    break
                    
                else:
                    return {"Método": "Bissecção", "Dados Iniciais": f"[{interv[0]},{interv[1]}]","Raíz x̅": f"{x:.8f}", "F(x̅)": f"{func(x):.13f}", "Erro em X": f"{error:.13f}", "N° de Iterações": k, "N° Lógico": logical_i, "Operações/Iteração": media_i, "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"}
                    # return x, k, func(x), (b-a)
            else:
                k += 1                         # Linha 9: Aumento da contagem do Iterador
                total_i.append(i)
                time_final = time.perf_counter()
                tempo_t.append(time_final - time_inicial) 
        else: 
            k += 1                             # Linha 9: Aumento da contagem do Iterador
            b = x                              # Linha 7: b = x
            i += 1
            total_i.append(i)
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial) 
                                               # Adição de contadores para o número de 
    return {"Método": "Bissecção", "Dados Iniciais": f"[{interv[0]},{interv[1]}]","Raíz x̅": f"{x:.8f}", "F(x̅)": f"{func(x):.13f}", "Erro em X": f"{(b-a):.13f}", "N° de Iterações": k, "N° Lógico": logical_i, "Operações/Iteração": round((sum(total_i))/k), "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"}
```

### Método 2: Método da Posição Falsa

#### Explicação do Método da Posição Falsa

Dada uma função $f(x)$ e um intervalo $[a,b]$, de forma que $f(a)\cdot f(b) < 0$.

Se pressupõe que existe apenas uma raíz no intervalo dado, então $f(x)=0$.  
O método da posição falsa não usa simplesmente uma média aritmética $\frac{a+b}{2}$, ele executa uma média aritmética ponderada entre $a$ e $b$ com os pesos $|f(a)|$ e $|f(b)|$, na seguinte forma:

$$ x = \frac{a \cdot |f(b)| + b \cdot |f(a)|}{|f(b)| + |f(a)|} = \frac {a \cdot f(b) - b \cdot f(a)}{f(b) - f(a)}$$

Complexidade do Métoda da Posição Falsa:  
Diferente da Bissecção, a Posição Falsa não possuí uma fórmula para estimar seu número de iterações, visto que a redução de seu intervalo não ocorre de maneira uniforme e depende da função que está sendo usada. Além do mais, se for exigido que dois critérios de parada sejam atendidos simultaneamente, o método pode exceder um número máximo de iterações.  
Então sua complexidade pode ser medida como o número de iterações obtidas ao fim da execução do método, além do fato de que o número de iterações vai depender da função a qual está sendo usada nele. 


Código para o Método da Posição Falsa:

```python
# Algoritmo 2 
# Código do Método da Posição Falsa

# função para teste
def funcao(x):
    return x**3 - x - 1                                                                                 
                                                                                     # Linha 1: Os dados iniciais                                              
interval = [1.0, 2.0]                                                                # O intervalo Inicial [a,b]
e1 = 10**(-6)
e2 = e1                                                                              # precisão ε

# função propriamente

def PosicaoFalsa(func, interval, e1, e2, p):
    k = 0
    a = interval[0]
    b = interval[1]
    total_i = []
    tempo_t = []
    logical_i = 0

    if abs(b-a) < e1:                                                                 # Linha 2: Condição para parada antes de iniciar a função, optei por usar o valor central entre o intervalo
        xi = (a+b)/2
        logical_i += 1
        return {"Método": "Secante", "Dados Iniciais": f"[{interval[0]}, {interval[1]}]", "Raíz x̅": f"{xi:.8f}", "F(x̅)": f"{func(xi):.13f}", "Erro em X": 0, "N° de Iterações": 0, "N° Lógico": logical_i, "Operações/Iteração": 0, "Total de Operações": 0, "Tempo Total": 0, "Tempo/Iteração": 0}
        # return xi

    k = 1                                                                            # Linha 3: Introdução de um iterador na função

    while abs(b-a) > e:
        time_inicial = time.perf_counter()
        i = 0

        M = func(a)                                                                  # Linha 4: variável para armazenar o valor de f(a) 
        x = (a*(func(b)) - b*(func(a))) / (func(b) - func(a))                        # Linha 5: Média ponderada da distância entre os intervalos 
        
        i += 11

        if abs(func(x)) < e2:
            i += 2
            total_i.append(i)
            media_i = round(sum(total_i)/k)
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
            error = abs(b-a)
            logical_i += 1
            if p == 1:
                resultados = [ ["Método", "Posição Falsa"], ["Dados Iniciais", f"[{interval[0]}, {interval[1]}]"], ["x̅ (raiz aproximada)", f"{x:.8f}"], ["f(x̅)", f"{func(x):.13f}"], ["Erro x̅", f"{error:.7f}"], ["Iterações", k], ["Operação(M)", media_i], ["Operação Total", sum(total_i)], ["Tempo Total", f"{sum(tempo_t):.12f}s"], ["Tempo por Iteração", f"{(sum(tempo_t) / k):.12f}s"]]
                print("\n" + "="*42+"\n          MÉTODO DA POSIÇÃO FALSA\n"+"="*42)
                print(tabulate(resultados, tablefmt="fancy_grid")) 
                print("="*42 + "\n")                                                 # O retorno é o x̅, o número de iterações, o valor da função em x e o erro em x(b-a)   
                break      
            else:
                return {"Método": "Posição Falsa", "Dados Iniciais": f"[{interval[0]}, {interval[1]}]","Raíz x̅": f"{x:.8f}", "F(x̅)": f"{func(x):.13f}", "Erro em X": f"{(b-a):.13f}", "N° de Iterações": k, "N° Lógico": logical_i, "Operações/Iteração": media_i, "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"}
                # return x, k, func(x), (b-a)       

        if (func(x)*M) > 0:                                                          # Linha 6: Condição para que a = x
            a = x 
            i += 4
            
            if (b - a) < e1:                                                         # Linha 8: Condição de parada
                xi = (a+b)/2
                i += 5                                                               # Print com Tabulate
                total_i.append(i)
                media_i = round(sum(total_i)/k)
                time_final = time.perf_counter()
                tempo_t.append(time_final - time_inicial)
                error = abs(b-a)
                logical_i += 1
                if p == 1:
                    resultados = [ ["Método", "Posição Falsa"], ["Dados Iniciais", f"[{interval[0]}, {interval[1]}]"], ["x̅ (raiz aproximada)", f"{x:.8f}"], ["f(x̅)", f"{func(x):.13f}"], ["Erro x̅", f"{error:.7f}"], ["Iterações", k], ["Operação(M)", media_i], ["Operação Total", sum(total_i)]]
                    print("\n" + "="*42+"\n          MÉTODO DA POSIÇÃO FALSA\n"+"="*42)
                    print(tabulate(resultados, tablefmt="fancy_grid")) 
                    print("="*42 + "\n")                                             # O retorno é o x̅, o número de iterações, o valor da função em x e o erro em x(b-a)   
                    break              
                else: 
                    return {"Método": "Secante", "Dados Iniciais": f"[{interval[0]},{interval[1]}]", "Raíz x̅": f"{x:.8f}", "F(x̅)": f"{func(x):.15f}", "Erro em X": f"{(b-a):.13f}", "N° de Iterações": k, "N° Lógico": logical_i, "Operações/Iteração": media_i, "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"}                      

            else:
                total_i.append(i)
                k += 1                                                               # Linha 9: Aumento da contagem do Iterador
                time_final = time.perf_counter()
                tempo_t.append(time_final - time_inicial)
        else: 
            k += 1                                                                   # Linha 9: Aumento da contagem do Iterador
            b = x 
            i += 1
            total_i.append(i)                                                                   # Linha 7: b = x
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
```

### Método 3: Método do Ponto Fixo(MPF)

#### Explicação do Método do Ponto Fixo(MPF)

Dada uma função contínua $f(x)$ num intervalo $[a,b]$, esse intervalo detém uma raíz da equação $f(x) = 0$.  
O método baseia-se em converter essa equação numa função equivalente $x = $ φ $(x)$ e por meio de um $x_0$ inicial gerar uma sequência 
$ \{x_k \}$ para $\xi$ usando a relação $x_{k+1} = $ φ$(x_{k})$, em razão da função φ$(x)$ ser tal que $f(\xi) = 0$ para que φ $(xi) = \xi $. Dessa forma o problema torna-se encontrar um ponto fixo em φ$(x)$.

</div>

Complexidade do Método do Ponto Fixo:


```python
# Algoritmo 3
def funcX(x):
    return x**3 - x - 1

def funcG(x):
    return (x+1)**(1/3)
                                                            # Linha 1: Os dados iniciais                                                    
x0 = 1                                                      # (a): Valor Inicial de x0
e1 = 10 **(-6)                                              # (b): Dados de Precisão


# Função Propriamente
def MPF(x_0, e1, e2, fx, fg, p):
    total_i = []
    dado_x = x_0
    tempo_t = []
    logical_i = 0

    if abs(fx(x_0)) < e1:                                   # Linha 2: Condição de parada inicial do MPF
        print(x_0)
        xi = x_0
        logical_i += 1
        return {"Método": "", "Dados Iniciais": f"X₀ = {dado_x}", "Raíz x̅": f"{xi:.8f}", "F(x̅)": f"{fx(xi):.13f}", "Erro em X": 0, "N° de Iterações": 0, "N° Lógico": logical_i, "Operações/Iteração": 0, "Total de Operações": 0, "Tempo Total": 0, "Tempo/Iteração": 0}
        # return xi, 0, fx(x_0), 0
    
    k = 1                                                   # Linha 3: Definição da variável de iteração 
    while abs(fx(x_0)) > e1:    
        time_inicial = time.perf_counter()
        i = 0
        x_1 = fg(x_0)                                       # Linha 4: Alteração do valor de x_1 baseado no valor de gx(x_0) 
        i += 4

        if abs(fx(x_1)) < e1 or abs(x_1 - x_0) < e2:        # Linha 5: Dada a condição x̅ = x_1
            xi = x_1
            i += 3                
            total_i.append(i)   
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
            error = abs(x_1 - x_0)
            logical_i += 1

            if p == 1:                                                                                                                                    # (abs(x_1 - x_0)) / abs(x_1)
                resultados = [ ["Método", "MPF"], ["Dados Iniciais", f"X₀ = {dado_x}"], ["x̅ (raiz aproximada)", f"{xi:.8f}"], ["f(x̅)", f"{fx(xi):.13f}"], ["Erro x̅", f"{error:.13f}"], ["Iterações", k], ["Operações(M)",  round((sum(total_i)/k))], ["Operação Total", sum(total_i)], ["Tempo Total", f"{sum(tempo_t):.12f}s"], ["Tempo por Iteração", f"{(sum(tempo_t) / k):.12f}s"]]
                print("\n" + "="*42+"\n               MÉTODO DA MPF\n"+"="*42)
                print(tabulate(resultados, tablefmt="fancy_grid")) 
                print("="*42 + "\n")        
                break
            else:  
                return {"Método": "MPF", "Dados Iniciais": f"X₀ = {dado_x}", "Raíz x̅": f"{xi:.8f}", "F(x̅)": f"{fx(xi):.15f}", "Erro em X": f"{error:.13f}", "N° de Iterações": k , "N° Lógico": logical_i, "Operações/Iteração": round((sum(total_i)/k)), "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"} 
                # return xi, k, fx(xi), (abs(x_1 - x_0)) / abs(x_1), round((sum(total_i)/k)), sum(total_i)
        
        else:
            k += 1                                          # Linha 6: Adição no iterador
            x_0 = x_1                                       # Linha 7: Atualização do valor de x_0
            i += 1
            total_i.append(i)
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
```

### Método 4: Método de Newton-Haphson

#### Explicação do Método de Newton-Haphson

Código do Método de Newton-Haphson

```python
# Algoritmo 4
def f(x):
    return x**3 - x - 1

def df(x):
    return 3*x**(2) - 1

                                                            # Linha 1: Dados iniciais
x0 = 0                                                      # (a): x_0 aproximação inicial
e1 = 10 ** (-6)                                             # (b): precisões e1 e e2


# Função propriamente
def NewtonHaphson(x_0, e1, e2, fx, dx, p):
    total_i = []
    tempo_t = []
    dado_x = x_0
    logical_i = 0

    if abs(fx(x_0)) < e1:                                  # Linha 2: Condição inicial de parada antes do loop
        xi = x_0
        logical_i += 1
        return {"Método": "", "Dados Iniciais": f"X₀ = {dado_x}", "Raíz x̅": f"{xi:.8f}", "F(x̅)": f"{fx(xi):.13f}", "Erro em X": 0, "N° de Iterações": 0, "N° Lógico": logical_i, "Operações/Iteração": 0, "Total de Operações": 0, "Tempo Total": 0, "Tempo/Iteração": 0}
        # return xi, 0, f(xi)
    
    k = 1                                                   # Linha 3: Inicialização do iterador

    while abs(fx(x_0)) > e1:
        time_inicial = time.perf_counter()
        i = 0
        x_1 = x_0 - (fx(x_0) / dx(x_0))                     # Linha 4: Atualização do valor de x_1 com o novo valor de x_0
        i += 5

        if abs(fx(x_1)) < e1 or abs(x_1 - x_0) < e2:        # Linha 5: Condição de parada usando e1 e e2 
            xi = x_1
            i += 3
            logical_i += 1
            total_i.append(i)
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)
            error = abs(x_1 - x_0)
            if p == 1:                                                                                                                                               # (abs(x_1 - x_0)) / abs(x_1)
                resultados = [ ["Método", "Newton-Haphson"], ["Dados Iniciais", f"X₀ = {dado_x}"], ["x̅ (raiz aproximada)", f"{xi:.8f}"], ["f(x̅)", f"{fx(xi):.15f}"], ["Erro x̅", f"{error:.13f}"], ["Iterações", k], ["Operações(M)",  round((sum(total_i)/k))], ["Operação Total", sum(total_i)], ["Tempo Total", f"{sum(tempo_t):.12f}s"], ["Tempo por Iteração", f"{(sum(tempo_t) / k):.12f}s"]]
                print("\n" + "="*42+"\n            MÉTODO NEWTON-HAPHSON\n"+"="*42)
                print(tabulate(resultados, tablefmt="fancy_grid")) 
                print("="*42 + "\n")        
                break    
            else: 
                return {"Método": "Newton-Haphson", "Dados Iniciais": f"X₀={dado_x}", "Raíz x̅": f"{xi:.8f}", "F(x̅)": f"{fx(xi):.15f}", "Erro em X": f"{error:.13f}", "N° de Iterações": k, "N° Lógico": logical_i, "Operações/Iteração": round((sum(total_i)/k)), "Total de Operações": sum(total_i), "Tempo Total": f"{sum(tempo_t):.12f}s", "Tempo/Iteração": f"{(sum(tempo_t)/k):.12f}s"}
                # return xi, k, fx(xi), (abs(x_1 - x_0)) / abs(x_1), round((sum(total_i)/k)), sum(total_i)
        else:
            x_0 = x_1                                        # Linha 6: Atualização do valor de x_0 com o valor obtido em x_1
            i += 1
            total_i.append(i)
            k += 1                                           # LInha 7: Atualização do iterador
            time_final = time.perf_counter()
            tempo_t.append(time_final - time_inicial)

    
```

### Adaptação do pseudo código acima para o método da Secante e do MPF:

#### Método para o valor de Polinômios:

```python
def polinomio(a, x):
    return sum([a[i] * (x ** i) for i in range(len(a))])
# É uma compressão de lista que por fim soma tudo internamente
# O 'i' vai de 0 até o tamanho da lista 'a' e a cada indíce é pego um valor da lista 'a' e
# o multiplica por (x ** i), após o laço a lista inteira é somada e retorna um valor 
# ele lê do maior grau para o menor
```

#### Método da Secante adaptado para o pseudo-código

```python
# Secante Atualizada

def Secante_Adaptada(a, x0, x1, e1, e2, itmax, fx):
    tempo_i = []

    # Avaliação inicial
    p0 = polinomio(a, x0)

    for k in range(1, itmax + 1):
        t_inicial = time.perf_counter()
        # b = p(x1)
        b = polinomio(a, x1)

        # Critério de parada 1
        if abs(b) <= e1:
            t_final = time.perf_counter()
            tempo_i.append(t_final - t_inicial)
            return {"Método": "Secante", "Raíz X": f"{x1:.12f}", "F(X)": f"{fx(x1):.12f}", "N° de Iterações": k, "Tempo/Iteração": f"{(sum(tempo_i)/k):.12f}", "Tempo Total": f"{sum(tempo_i):.12f}"}

        # c = derivada aproximada (secante)
        c = (b - p0) / (x1 - x0)

        # Correção (passo da secante)
        deltax = b / c
        x2 = x1 - deltax

        # Critério de parada 2
        if abs(deltax) <= e2:
            return {"Método": "Secante", "Raíz X": f"{x2:.12f}", "F(X)": f"{fx(x2):.12f}", "N° de Iterações": k, "Tempo/Iteração": f"{(sum(tempo_i)/k):.12f}", "Tempo Total": f"{sum(tempo_i):.12f}"}

        #
        x0 = x1
        p0 = b
        x1 = x2
        t_final = time.perf_counter()
        tempo_i.append(t_final - t_inicial)
    print(f"O método da Secante não convergiu com {itmax} iterações")
```

#### Método do Ponto Fixo adaptado para o pseudo-código

```python
def mpf_polinomio(a, x0, e1, e2, itmax, g):
    tempo_i = []
    x = x0

    for k in range(1, itmax + 1):
        t_inicial = time.perf_counter()
        # b = P(x)
        b = polinomio(a, x)

        # Critério 1: valor da função
        if abs(b) <= e1:
            t_final = time.perf_counter()
            tempo_i.append(t_final - t_inicial)
            
            return {"Método": "Ponto Fixo", "Raíz X": f"{x:.12f}", "F(X)": f"{b:.12f}", "N° de Iterações": k, "Tempo/Iteração": f"{(sum(tempo_i)/k):.12f}", "Tempo Total": f"{sum(tempo_i):.12f}"}

        # Iteração do ponto fixo
        x_new = g(a, x)

        # deltax = x_{k+1} - x_k
        deltax = x_new - x

        # Atualização
        x = x_new

        # Critério 2: estabilidade
        if abs(deltax) <= e2:
            t_final = time.perf_counter()
            tempo_i.append(t_final - t_inicial)
            return {"Método": "Ponto Fixo", "Raíz X": f"{x:.12f}", "F(X)": f"{polinomio(a, x):.12f}", "N° de Iterações": k, "Tempo/Iteração": f"{(sum(tempo_i)/k):.12f}", "Tempo Total": f"{sum(tempo_i):.12f}"}

    print(f"O método da Secante não convergiu com {itmax} iterações")
```
